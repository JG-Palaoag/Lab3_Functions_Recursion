```python
# [CELL 1] STUDENT CONFIGURATION

LAST_NAME = "PALAOAG"
SEED_NUM = 9
FAVORITE_ARTIST = "RICO BLANCO"

ARTIST_NAME_LENGTH = len(FAVORITE_ARTIST)
NAME_LENGTH = len(LAST_NAME)

print(LAST_NAME)
print(SEED_NUM)
print(FAVORITE_ARTIST)
print(ARTIST_NAME_LENGTH)
print(NAME_LENGTH)
```

```python
# [CELL 2] Exercise 1: Equipment Diagnostic System

# Equipment Data

EQUIPMENT_DATA = (SEED_NUM + NAME_LENGTH + ARTIST_NAME_LENGTH) * 3

def validate_readings(data):
    validation_results = {}

    limits = {
        "Validation Results": (0, 100)
    }

    for key, limits in limits.items():
        if data < limits[0] or data > limits[1]:
            validation_results[key] = "Invalid"
        else:
            validation_results[key] = "Valid"

    return validation_results

execution_log = []

def diagnostic_logger(func):
    def wrapper(*args, **kwargs):
        execution_log.append(f"Starting diagnostic: {func.__name__}")

        try:
            result = func(*args, **kwargs)
            execution_log.append(f"Completed diagnostic: {func.__name__}")
            return result

        except Exception as e:
            execution_log.append(f"Error in {func.__name__}: {e}")
            raise

    return wrapper

@diagnostic_logger
def run_diagnostic(data):
    validation = validate_readings(data)

    condition = assign_condition(data)

    return validation, condition

def assign_condition(EQUIPMENT_DATA):
    if EQUIPMENT_DATA >= 90:
        return 'Excellent Condition'
    elif EQUIPMENT_DATA >= 80:
        return 'Good Condition'
    elif EQUIPMENT_DATA >= 70:
        return 'Fair Condition'
    elif EQUIPMENT_DATA >= 60: 
        return 'Poor Condition'
    else:
        return 'Critical Condition'

validation_results, condition = run_diagnostic(EQUIPMENT_DATA)

print("Equipment Diagnostic System")
print("=" * 40)
print("Equipment Data:", EQUIPMENT_DATA)
validation_results = validate_readings(EQUIPMENT_DATA)
for key, value in validation_results.items():
    print(f"{key}: {value}")
print("Diagnostic Results:", assign_condition(EQUIPMENT_DATA))
print("Execution Log:")
for log in execution_log:
    print(log)
print("Final Output:")
print(
    f"Condition:", assign_condition(EQUIPMENT_DATA)
)
print(
    f"Data:", (EQUIPMENT_DATA)
)
```

```python
# [CELL 3] Exercise 2

def generate_fault_code(last_name, seed_num, artist):
    name_value = sum(ord(char) for char in last_name)
    artist_value = sum(ord(char) for char in artist)

    fault_code = (name_value + artist_value + seed_num) % 1000

    return fault_code


def trace_fault(code, trace=None):

    if trace is None:
        trace = []

    trace.append(code)

    
    if code <= 10:
        return trace

    next_code = code // 2

    return trace_fault(next_code, trace)


fault_code = generate_fault_code(
    LAST_NAME,
    SEED_NUM,
    FAVORITE_ARTIST
)

fault_trace = trace_fault(fault_code)

recursive_calls = len(fault_trace)

print("RECURSIVE FAULT TRACE")
print("=" * 50)
print(f"Generated Fault Code: {fault_code}")

print("Recursive Trace:")
print(" -> ".join(map(str, fault_trace)))

print(f"Number of Recursive Calls: {recursive_calls}")

print(
    f"Final Result: "
    f"Fault trace terminated at {fault_trace[-1]}"
)
```

```python
# [CELL 4] Exercise 3

def generate_telemetry(last_name, seed_num, artist, count=12):

    name_value = sum(ord(char) for char in last_name)
    artist_value = sum(ord(char) for char in artist)

    base_value = (
        name_value
        + seed_num * 10
        + artist_value
    ) % 60

    for i in range(count):

        value = (
            base_value
            + i * 19
            + len(artist) * 3
        ) % 121 - 10

        yield value

def validate_reading(value):

    if value < 0 or value > 100:
        raise ValueError(
            f"Invalid telemetry value: {value}"
        )

    return True

process_reading = lambda value: round(
    value * 1.1,
    1
)


def monitor_function(function):

    def wrapper(*args, **kwargs):

        wrapper.call_count += 1

        return function(*args, **kwargs)

    wrapper.call_count = 0

    return wrapper

@monitor_function
def process_valid_reading(value):

    return process_reading(value)


def analyze_abnormal(value, trace=None):

    if trace is None:
        trace = []

    trace.append(round(value, 1))

    # BASE CONDITION
    if 15 <= value <= 90:
        return trace

    # HIGH ABNORMAL VALUE
    if value > 90:
        return analyze_abnormal(
            value - 10,
            trace
        )

    # LOW ABNORMAL VALUE
    return analyze_abnormal(
        value + 5,
        trace
    )

def run_monitoring(telemetry_stream):

    valid_count = 0
    invalid_count = 0

    processed_values = []
    abnormal_conditions = []

    for reading in telemetry_stream:

        print(f"Incoming telemetry: {reading}")

        try:
            validate_reading(reading)

            valid_count += 1

            processed = process_valid_reading(
                reading
            )

            processed_values.append(processed)

            print(
                f"  Valid -> Processed: {processed}"
            )

            if processed > 90 or processed < 15:

                trace = analyze_abnormal(
                    processed
                )

                abnormal_conditions.append({
                    "value": processed,
                    "trace": trace
                })

                print(
                    f"  ABNORMAL -> "
                    f"Recursive trace: {trace}"
                )

        except ValueError as error:

            invalid_count += 1

            print(
                f"  INVALID -> {error}"
            )

    if (
        invalid_count > 0
        or len(abnormal_conditions) > 0
    ):

        status = "ATTENTION REQUIRED"

    else:

        status = "NORMAL"

    return {
        "processed_count": valid_count,
        "valid_count": valid_count,
        "invalid_count": invalid_count,
        "processed_values": processed_values,
        "abnormal_conditions": abnormal_conditions,
        "status": status,
        "decorator_calls":
            process_valid_reading.call_count
    }

print("INTELLIGENT EQUIPMENT MONITORING PIPELINE")
print("=" * 45)

telemetry_stream = generate_telemetry(
    LAST_NAME,
    SEED_NUM,
    FAVORITE_ARTIST,
    count=12
)


print("\n--- TELEMETRY PROCESSING ---")


results = run_monitoring(
    telemetry_stream
)

print("FINAL DIAGNOSTIC REPORT")
print("=" * 45)
print(
    f"Processed Readings: "
    f"{results['processed_count']}"
)

print(
    f"Valid Readings: "
    f"{results['valid_count']}"
)

print(
    f"Invalid Readings: "
    f"{results['invalid_count']}"
)

print("Processed Results:")
print(
    results["processed_values"]
)

print("Detected Abnormal Conditions:")
if results["abnormal_conditions"]:

    for abnormal in results["abnormal_conditions"]:

        print(
            f"Value {abnormal['value']} "
            f"-> Trace: {abnormal['trace']}"
        )

else:

    print("None")

print(
    f"\nDecorator Processing Calls: "
    f"{results['decorator_calls']}"
)
print(
    f"Overall Equipment Status: "
    f"{results['status']}"
)
print("Monitoring complete.")
```
